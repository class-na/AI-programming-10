## 화면에 글자 내보내기

- print는 '화면에 보여 줘'라는 명령입니다.

**실행 결과:**

````python
print("안녕하세요")
````

## 컴퓨터는 시킨 대로만 합니다

- 쓰는 방법이 틀리면 아예 실행되지 않습니다.
- 쓰는 방법이 맞아도 뜻이 틀릴 수 있습니다.
- 앞의 것이 문법, 뒤의 것이 의미입니다.
- 둘 다 결국 사람이 확인해야 합니다.

> 📖 **문법(syntax)** — 코드를 쓰는 방법에 관한 약속. 맞춤법과 같습니다.
> 📖 **의미(semantics)** — 그 코드가 실제로 하는 일. 문장의 뜻과 같습니다.

- 순서를 바꾸려면 괄호를 씁니다.

## 프로그램이 만들어지는 순서

- 1) 하고 싶은 일을 정합니다.
- 2) 그 일을 잘게 순서로 나눕니다.
- 3) 순서를 파이썬 문장으로 옮깁니다.
- 4) 실행해서 결과를 봅니다.
- 5) 틀렸으면 고쳐서 다시 실행합니다.

## 파이썬은 한 줄씩 바로 실행합니다

- 적자마자 바로 돌려 볼 수 있습니다.
- 틀린 곳은 그 줄에 갔을 때 알려 줍니다.
- 그래서 조금씩 쓰고 자주 실행하는 게 좋습니다.
- 우리 수업은 이 방식으로 진행합니다.

> 📖 **인터프리터(interpreter)** — 코드를 한 줄씩 읽어 바로 실행해 주는 프로그램. 파이썬이 이 방식입니다.

## 오류는 세 가지가 있습니다

- 1) **문법 오류** — 실행조차 안 됩니다.
- 2) **실행 중 오류** — 가다가 멈춥니다.
- 3) **논리 오류** — 끝까지 도는데 답이 틀립니다.
- 셋 중 무엇인지 아는 것이 첫걸음입니다.

## AI가 잘하는 것 / 조심할 것

- 잘함 — 오류 메시지를 쉬운 말로 풀어 줍니다.
- 잘함 — 어디를 고쳐야 하는지 짚어 줍니다.
- 조심 — 틀린 말도 자신 있게 합니다.
- 조심 — 받은 답은 반드시 실행해 봅니다.

> 📖 **프롬프트(prompt)** — AI에게 주는 질문이나 지시. 자세할수록 답이 좋아집니다.
> 📖 **환각(hallucination)** — AI가 사실이 아닌 내용을 사실처럼 말하는 것.
